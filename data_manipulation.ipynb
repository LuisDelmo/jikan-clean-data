```python
import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


base_url = 'https://api.jikan.moe/v4/anime'

params = {
    'unapproved': 'false',
    'q': 'jujutsu kaisen'
}

response = requests.get(url=base_url,params=params)

if response.status_code != 200:
    raise ValueError(response.status_code)

full_data = response.json()

clean_data = full_data['data']

df = pd.json_normalize(clean_data)
```

```python
df.head()
```

```python
check_title_juju = df['title_english'].str.contains('jujutsu', case=False, na=False)
df_juju = df[check_title_juju]
```

```python
df_juju
```

```python


plt.bar(x=df_juju['title_english'],height=df_juju['score'])
plt.xticks()
plt.show()
```
